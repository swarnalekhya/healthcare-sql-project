```python
# ============================================================
# CELL 1 — Setup: import libraries and create the database
# Run this first every time. Takes about 2 seconds.
# ============================================================

import sqlite3
import pandas as pd

# Makes tables display nicely
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
pd.set_option('display.width', 120)

# Create database in memory (or change ':memory:' to 'healthcare.db' to save to disk)
conn = sqlite3.connect(':memory:')

# Helper function — run any SQL and see results as a table
def run(sql):
    return pd.read_sql_query(sql, conn)

print('Libraries loaded. Database ready.')
```

```python
# ============================================================
# CELL 2 — Create tables
# ============================================================

conn.executescript("""
    DROP TABLE IF EXISTS diagnoses;
    DROP TABLE IF EXISTS admissions;
    DROP TABLE IF EXISTS patients;

    CREATE TABLE patients (
        patient_id      INTEGER PRIMARY KEY,
        first_name      TEXT,
        last_name       TEXT,
        date_of_birth   TEXT,
        gender          TEXT,
        zip_code        TEXT,
        insurance_type  TEXT
    );

    CREATE TABLE admissions (
        admission_id    INTEGER PRIMARY KEY,
        patient_id      INTEGER,
        admission_date  TEXT,
        discharge_date  TEXT,
        department      TEXT,
        admission_type  TEXT,
        attending_md    TEXT
    );

    CREATE TABLE diagnoses (
        diagnosis_id            INTEGER PRIMARY KEY,
        admission_id            INTEGER,
        patient_id              INTEGER,
        icd_code                TEXT,
        diagnosis_description   TEXT,
        diagnosis_type          TEXT
    );
""")

print('Tables created successfully.')
```

## Queries
Run each cell below independently. Results display as a table automatically.

```python
# ============================================================
# QUERY 1 — Total admissions by department
# Concept: Basic GROUP BY + ORDER BY
# ============================================================

run("""
    SELECT
        department,
        COUNT(*) AS total_admissions
    FROM admissions
    GROUP BY department
    ORDER BY total_admissions DESC
""")
```

```python
# ============================================================
# QUERY 2 — Average length of stay by department
# Concept: Date math with JULIANDAY (SQLite version of DATEDIFF)
# ============================================================

run("""
    SELECT
        department,
        COUNT(*)                                                            AS total_admissions,
        ROUND(AVG(JULIANDAY(discharge_date) - JULIANDAY(admission_date)),1) AS avg_los_days,
        MIN(JULIANDAY(discharge_date) - JULIANDAY(admission_date))          AS min_los_days,
        MAX(JULIANDAY(discharge_date) - JULIANDAY(admission_date))          AS max_los_days
    FROM admissions
    GROUP BY department
    ORDER BY avg_los_days DESC
""")
```

```python
# ============================================================
# QUERY 3 — Admissions by insurance type
# Concept: JOIN across two tables
# ============================================================

run("""
    SELECT
        p.insurance_type,
        COUNT(a.admission_id)        AS total_admissions,
        COUNT(DISTINCT a.patient_id) AS unique_patients
    FROM admissions a
    JOIN patients p ON a.patient_id = p.patient_id
    GROUP BY p.insurance_type
    ORDER BY total_admissions DESC
""")
```

```python
# ============================================================
# QUERY 4 — Emergency vs elective vs urgent per department
# Concept: CASE WHEN conditional aggregation
# ============================================================

run("""
    SELECT
        department,
        COUNT(*)                                                              AS total,
        SUM(CASE WHEN admission_type = 'Emergency' THEN 1 ELSE 0 END)        AS emergency,
        SUM(CASE WHEN admission_type = 'Elective'  THEN 1 ELSE 0 END)        AS elective,
        SUM(CASE WHEN admission_type = 'Urgent'    THEN 1 ELSE 0 END)        AS urgent,
        ROUND(
            100.0 * SUM(CASE WHEN admission_type = 'Emergency' THEN 1 ELSE 0 END) / COUNT(*)
        , 1)                                                                  AS emergency_pct
    FROM admissions
    GROUP BY department
    ORDER BY emergency_pct DESC
""")
```

```python
# ============================================================
# QUERY 5 — Top primary diagnoses
# Concept: Filter on ICD type, count across patients
# ============================================================

run("""
    SELECT
        icd_code,
        diagnosis_description,
        COUNT(*)                     AS diagnosis_count,
        COUNT(DISTINCT patient_id)   AS unique_patients
    FROM diagnoses
    WHERE diagnosis_type = 'Primary'
    GROUP BY icd_code, diagnosis_description
    ORDER BY diagnosis_count DESC
    LIMIT 10
""")
```

```python
# ============================================================
# QUERY 6 — 30-day readmission events
# Concept: Self JOIN — compare each visit to future visits
# ============================================================

run("""
    SELECT
        a1.patient_id,
        p.first_name || ' ' || p.last_name             AS patient_name,
        a1.department,
        a1.discharge_date                               AS first_discharge,
        a2.admission_date                               AS readmit_date,
        CAST(
            JULIANDAY(a2.admission_date) - JULIANDAY(a1.discharge_date)
        AS INTEGER)                                     AS days_between
    FROM admissions a1
    JOIN admissions a2
        ON  a1.patient_id = a2.patient_id
        AND a2.admission_date > a1.discharge_date
        AND (JULIANDAY(a2.admission_date) - JULIANDAY(a1.discharge_date)) <= 30
    JOIN patients p ON a1.patient_id = p.patient_id
    ORDER BY a1.patient_id, a1.discharge_date
""")
```

```python
# ============================================================
# QUERY 7 — 30-day readmission RATE by department
# Concept: CTE (WITH clause) + LEFT JOIN
# ============================================================

run("""
    WITH readmits AS (
        SELECT DISTINCT
            a1.patient_id,
            a1.department,
            a1.discharge_date
        FROM admissions a1
        JOIN admissions a2
            ON  a1.patient_id = a2.patient_id
            AND a2.admission_date > a1.discharge_date
            AND (JULIANDAY(a2.admission_date) - JULIANDAY(a1.discharge_date)) <= 30
    )
    SELECT
        a.department,
        COUNT(*)                                                    AS total_discharges,
        COUNT(r.patient_id)                                         AS readmissions,
        ROUND(100.0 * COUNT(r.patient_id) / COUNT(*), 1)           AS readmit_rate_pct
    FROM admissions a
    LEFT JOIN readmits r
        ON  a.patient_id     = r.patient_id
        AND a.department     = r.department
        AND a.discharge_date = r.discharge_date
    GROUP BY a.department
    ORDER BY readmit_rate_pct DESC
""")
```

```python
# ============================================================
# QUERY 8 — High-risk patients: 3+ admissions AND diabetic
# Concept: Multi-table join + HAVING + ICD code LIKE filter
# ============================================================

run("""
    SELECT
        a.patient_id,
        p.first_name || ' ' || p.last_name             AS patient_name,
        p.insurance_type,
        COUNT(DISTINCT a.admission_id)                  AS total_admissions,
        ROUND(
            AVG(JULIANDAY(a.discharge_date) - JULIANDAY(a.admission_date))
        , 1)                                            AS avg_los
    FROM admissions a
    JOIN patients p   ON a.patient_id = p.patient_id
    JOIN diagnoses d  ON a.patient_id = d.patient_id
    WHERE d.icd_code LIKE 'E11%'
    GROUP BY a.patient_id, p.first_name, p.last_name, p.insurance_type
    HAVING COUNT(DISTINCT a.admission_id) >= 3
    ORDER BY total_admissions DESC
""")
```

```python
# ============================================================
# QUERY 9 — Patient visit timeline with window functions
# Concept: ROW_NUMBER and LAG — senior-level SQL
# SQLite supports window functions from version 3.25 (2018)
# ============================================================

run("""
    SELECT
        a.patient_id,
        p.first_name || ' ' || p.last_name             AS patient_name,
        a.admission_date,
        a.discharge_date,
        a.department,
        CAST(
            JULIANDAY(a.discharge_date) - JULIANDAY(a.admission_date)
        AS INTEGER)                                     AS los_days,
        ROW_NUMBER() OVER (
            PARTITION BY a.patient_id
            ORDER BY a.admission_date
        )                                               AS visit_number,
        LAG(a.discharge_date) OVER (
            PARTITION BY a.patient_id
            ORDER BY a.admission_date
        )                                               AS prev_discharge,
        CAST(
            JULIANDAY(a.admission_date) -
            JULIANDAY(LAG(a.discharge_date) OVER (
                PARTITION BY a.patient_id ORDER BY a.admission_date
            ))
        AS INTEGER)                                     AS days_since_last_discharge
    FROM admissions a
    JOIN patients p ON a.patient_id = p.patient_id
    ORDER BY a.patient_id, a.admission_date
""")
```

```python
# ============================================================
# QUERY 10 — Full patient summary report
# Concept: Multiple CTEs + everything combined
# This is what you'd hand to a clinical director
# ============================================================

run("""
    WITH patient_stats AS (
        SELECT
            patient_id,
            COUNT(DISTINCT admission_id)                                            AS total_admissions,
            ROUND(AVG(JULIANDAY(discharge_date) - JULIANDAY(admission_date)), 1)   AS avg_los,
            SUM(CASE WHEN admission_type = 'Emergency' THEN 1 ELSE 0 END)          AS emergency_visits,
            MAX(admission_date)                                                     AS last_admission
        FROM admissions
        GROUP BY patient_id
    ),
    readmit_flags AS (
        SELECT DISTINCT a1.patient_id
        FROM admissions a1
        JOIN admissions a2
            ON  a1.patient_id = a2.patient_id
            AND a2.admission_date > a1.discharge_date
            AND (JULIANDAY(a2.admission_date) - JULIANDAY(a1.discharge_date)) <= 30
    ),
    diabetes_flags AS (
        SELECT DISTINCT patient_id
        FROM diagnoses
        WHERE icd_code LIKE 'E11%'
    )
    SELECT
        p.patient_id,
        p.first_name || ' ' || p.last_name             AS patient_name,
        p.gender,
        p.insurance_type,
        ps.total_admissions,
        ps.avg_los                                      AS avg_los_days,
        ps.emergency_visits,
        ps.last_admission,
        CASE WHEN rf.patient_id IS NOT NULL THEN 'Yes' ELSE 'No' END AS had_readmission,
        CASE WHEN df.patient_id IS NOT NULL THEN 'Yes' ELSE 'No' END AS has_diabetes
    FROM patients p
    JOIN patient_stats ps      ON p.patient_id = ps.patient_id
    LEFT JOIN readmit_flags rf ON p.patient_id = rf.patient_id
    LEFT JOIN diabetes_flags df ON p.patient_id = df.patient_id
    ORDER BY ps.total_admissions DESC, ps.avg_los DESC
""")
```

```python
# ============================================================
# QUERY 11 — High-Risk Patient Flagging (Multi-Factor Risk Score)
# Concept: CASE WHEN with multiple conditions + GROUP_CONCAT
# ============================================================
 
run("""
    SELECT
        p.patient_id,
        p.first_name || ' ' || p.last_name             AS patient_name,
        p.insurance_type,
        COUNT(DISTINCT a.admission_id)                  AS total_admissions,
        ROUND(AVG(
            JULIANDAY(a.discharge_date) - JULIANDAY(a.admission_date)
        ), 1)                                           AS avg_los,
        GROUP_CONCAT(DISTINCT d.icd_code)               AS diagnoses,
        CASE
            WHEN COUNT(DISTINCT a.admission_id) > 1
             AND MAX(CASE
                    WHEN d.icd_code IN ('I50.9','J44.1','J18.9','I21.0','I21.9')
                    THEN 1 ELSE 0
                 END) = 1
            THEN 'High Risk'
            ELSE 'Standard'
        END                                             AS risk_category
    FROM patients p
    JOIN admissions a ON p.patient_id = a.patient_id
    JOIN diagnoses d  ON a.admission_id = d.admission_id
    GROUP BY p.patient_id, p.first_name, p.last_name, p.insurance_type
    ORDER BY total_admissions DESC
""")
```

```python
#============================================================
# QUERY 12 — Readmission Rate by Diagnosis
# Concept: CTE to calculate readmission flag, then join to diagnoses
# ============================================================
 
run("""
    WITH readmit_events AS (
        SELECT DISTINCT a1.admission_id
        FROM admissions a1
        JOIN admissions a2
            ON  a1.patient_id = a2.patient_id
            AND a2.admission_date > a1.discharge_date
            AND (JULIANDAY(a2.admission_date) - JULIANDAY(a1.discharge_date)) <= 30
    )
    SELECT
        d.icd_code,
        d.diagnosis_description,
        d.diagnosis_type,
        COUNT(DISTINCT d.admission_id)                  AS total_admissions,
        COUNT(DISTINCT r.admission_id)                  AS readmissions,
        ROUND(
            100.0 * COUNT(DISTINCT r.admission_id) /
            COUNT(DISTINCT d.admission_id)
        , 1)                                            AS readmit_rate_pct
    FROM diagnoses d
    LEFT JOIN readmit_events r ON d.admission_id = r.admission_id
    WHERE d.diagnosis_type = 'Primary'
    GROUP BY d.icd_code, d.diagnosis_description, d.diagnosis_type
    HAVING COUNT(DISTINCT d.admission_id) >= 2
    ORDER BY readmit_rate_pct DESC
""")
```

```python
# ============================================================
# QUERY 13 — Department Performance Scorecard
# Concept: Everything in one view — the "dashboard in a query"
# Multiple KPIs in a single clean executive summary.
# ============================================================
 
run("""
    WITH readmit_events AS (
        SELECT DISTINCT a1.admission_id
        FROM admissions a1
        JOIN admissions a2
            ON  a1.patient_id = a2.patient_id
            AND a2.admission_date > a1.discharge_date
            AND (JULIANDAY(a2.admission_date) - JULIANDAY(a1.discharge_date)) <= 30
    )
    SELECT
        a.department,
        COUNT(DISTINCT a.admission_id)                  AS total_admissions,
        COUNT(DISTINCT a.patient_id)                    AS unique_patients,
        ROUND(AVG(
            JULIANDAY(a.discharge_date) - JULIANDAY(a.admission_date)
        ), 1)                                           AS avg_los_days,
        SUM(CASE WHEN a.admission_type = 'Emergency'
            THEN 1 ELSE 0 END)                          AS emergency_count,
        ROUND(100.0 * SUM(CASE WHEN a.admission_type = 'Emergency'
            THEN 1 ELSE 0 END) / COUNT(*), 1)           AS emergency_pct,
        COUNT(DISTINCT r.admission_id)                  AS readmissions,
        ROUND(
            100.0 * COUNT(DISTINCT r.admission_id) /
            COUNT(DISTINCT a.admission_id)
        , 1)                                            AS readmit_rate_pct
    FROM admissions a
    LEFT JOIN readmit_events r ON a.admission_id = r.admission_id
    GROUP BY a.department
    ORDER BY readmit_rate_pct DESC
""")
```

```python
# ============================================================
# BONUS — Bar chart: 30-day readmission rate by department
# ============================================================

df2 = run("""
    WITH readmits AS (
        SELECT DISTINCT a1.patient_id, a1.department, a1.discharge_date
        FROM admissions a1
        JOIN admissions a2
            ON  a1.patient_id = a2.patient_id
            AND a2.admission_date > a1.discharge_date
            AND (JULIANDAY(a2.admission_date) - JULIANDAY(a1.discharge_date)) <= 30
    )
    SELECT
        a.department,
        ROUND(100.0 * COUNT(r.patient_id) / COUNT(*), 1) AS readmit_rate_pct
    FROM admissions a
    LEFT JOIN readmits r
        ON a.patient_id = r.patient_id
        AND a.department = r.department
        AND a.discharge_date = r.discharge_date
    GROUP BY a.department
    ORDER BY readmit_rate_pct DESC
""")

fig, ax = plt.subplots(figsize=(9, 5))
colors = ['#D85A30' if x >= 30 else '#1D9E75' for x in df2['readmit_rate_pct']]
bars = ax.barh(df2['department'], df2['readmit_rate_pct'], color=colors, edgecolor='none')
ax.set_xlabel('30-Day Readmission Rate (%)', fontsize=11)
ax.set_title('30-Day Readmission Rate by Department — 2024', fontsize=13, fontweight='bold', pad=14)
ax.bar_label(bars, fmt='%.1f%%', padding=4, fontsize=10)
ax.invert_yaxis()
ax.spines[['top','right','left']].set_visible(False)
ax.tick_params(left=False)
ax.set_xlim(0, df2['readmit_rate_pct'].max() + 10)
plt.tight_layout()
plt.savefig('readmission_rate.png', dpi=150, bbox_inches='tight')
plt.show()
print('Chart saved as readmission_rate.png')
```

```python
#============================================================
# VISUALIZATION — Comorbidity Heatmap
# Which diagnoses appear together most often in the same patient?
# Concept: Co-occurrence matrix — shows clinical pattern recognition
# This is the kind of analysis used in population health and
# chronic disease management programs
# ============================================================
 
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
from itertools import combinations
 
# Step 1 — Get all unique diagnoses per patient
diag_df = run("""
    SELECT DISTINCT
        patient_id,
        icd_code,
        diagnosis_description
    FROM diagnoses
    ORDER BY patient_id, icd_code
""")
 
# Step 2 — Short readable labels for each ICD code
icd_labels = {
    'I21.0':  'Acute MI\n(Anterior)',
    'I21.9':  'Acute MI\n(Unspec)',
    'I50.9':  'Heart\nFailure',
    'I10':    'Hyper-\ntension',
    'I63.9':  'Stroke',
    'G45.9':  'TIA',
    'J18.9':  'Pneumonia',
    'J44.1':  'COPD',
    'E11.9':  'T2 Diabetes\n(Uncontrolled)',
    'E11.65': 'T2 Diabetes\n(Hypoglycemia)',
    'M17.11': 'Osteo-\narthritis',
}
 
# Step 3 — Build co-occurrence matrix
# For each patient, find all pairs of diagnoses that appear together
icd_codes = sorted(diag_df['icd_code'].unique())
n = len(icd_codes)
matrix = np.zeros((n, n), dtype=int)
idx = {code: i for i, code in enumerate(icd_codes)}
 
patient_groups = diag_df.groupby('patient_id')['icd_code'].apply(list)
 
for codes in patient_groups:
    unique_codes = list(set(codes))
    # Diagonal — count how many patients have each diagnosis
    for c in unique_codes:
        matrix[idx[c]][idx[c]] += 1
    # Off-diagonal — count co-occurrences
    for c1, c2 in combinations(unique_codes, 2):
        matrix[idx[c1]][idx[c2]] += 1
        matrix[idx[c2]][idx[c1]] += 1
 
# Step 4 — Build labels
labels = [icd_labels.get(c, c) for c in icd_codes]
 
# Step 5 — Mask the diagonal so co-occurrence pairs stand out
mask = np.eye(n, dtype=bool)
display_matrix = np.where(mask, np.nan, matrix.astype(float))
 
# Step 6 — Plot
fig, ax = plt.subplots(figsize=(13, 10))
 
# Custom colormap — white for zero, teal gradient for co-occurrences
import matplotlib.colors as mcolors
cmap = mcolors.LinearSegmentedColormap.from_list(
    'healthcare',
    ['#F7FAFA', '#A8DDD4', '#4DB6A8', '#1D9E75', '#0F6E56'],
    N=256
)
cmap.set_bad('#F0F0F0')  # diagonal cells in light grey
 
im = ax.imshow(display_matrix, cmap=cmap, aspect='auto', vmin=0)
 
# Diagonal styling — grey boxes
for i in range(n):
    ax.add_patch(mpatches.FancyBboxPatch(
        (i - 0.5, i - 0.5), 1, 1,
        boxstyle='square,pad=0',
        facecolor='#E8E8E8',
        edgecolor='white',
        linewidth=0.5,
        zorder=2
    ))
    ax.text(i, i, str(matrix[i][i]),
            ha='center', va='center',
            fontsize=9, color='#666666',
            fontweight='500', zorder=3)
 
# Annotate co-occurrence cells
for i in range(n):
    for j in range(n):
        if i != j and matrix[i][j] > 0:
            ax.text(j, i, str(matrix[i][j]),
                    ha='center', va='center',
                    fontsize=10,
                    color='white' if matrix[i][j] >= 3 else '#1D4A40',
                    fontweight='bold' if matrix[i][j] >= 3 else 'normal')
 
# Axes formatting
ax.set_xticks(range(n))
ax.set_yticks(range(n))
ax.set_xticklabels(labels, fontsize=9, rotation=45, ha='right')
ax.set_yticklabels(labels, fontsize=9)
ax.tick_params(length=0)
 
# Grid lines
for i in range(n + 1):
    ax.axhline(i - 0.5, color='white', linewidth=1.5)
    ax.axvline(i - 0.5, color='white', linewidth=1.5)
 
# Colorbar
cbar = plt.colorbar(im, ax=ax, shrink=0.6, pad=0.02)
cbar.set_label('Co-occurrence count\n(patients sharing both diagnoses)',
               fontsize=9, labelpad=10)
cbar.ax.tick_params(labelsize=8)
 
# Titles and labels
ax.set_title(
    'Diagnosis Comorbidity Heatmap — Patient Co-occurrence Patterns',
    fontsize=14, fontweight='bold', pad=20
)
ax.set_xlabel('Diagnosis', fontsize=11, labelpad=12)
ax.set_ylabel('Diagnosis', fontsize=11, labelpad=12)
 
# Annotation box explaining the chart
ax.annotate(
    'Grey diagonal = patients with that diagnosis\n'
    'Coloured cells = patients sharing both diagnoses',
    xy=(0.01, -0.22), xycoords='axes fraction',
    fontsize=9, color='#555555',
    bbox=dict(boxstyle='round,pad=0.4', facecolor='#F5F5F5', edgecolor='#CCCCCC')
)
 
plt.tight_layout()
plt.savefig('comorbidity_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved: comorbidity_heatmap.png')
 
# Step 7 — Print top comorbidity pairs as text findings
print('\n--- TOP COMORBIDITY PAIRS ---')
pairs = []
for i in range(n):
    for j in range(i+1, n):
        if matrix[i][j] > 0:
            pairs.append((matrix[i][j], icd_codes[i], icd_codes[j]))
pairs.sort(reverse=True)
for count, c1, c2 in pairs[:5]:
    print(f"  {count} patient(s): {icd_labels.get(c1,c1).replace(chr(10),' ')} + {icd_labels.get(c2,c2).replace(chr(10),' ')}")
 
```

```python
# ============================================================
# VISUALIZATION — Patient Risk Scatter Plot (Bubble Chart)
# X axis = total admissions
# Y axis = average length of stay
# Bubble size = number of unique diagnoses
# Color = risk category (High Risk vs Standard)
# Outlier patients appear in the top-right corner
# ============================================================
 
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
 
# Pull patient-level stats
risk_df = run("""
    WITH patient_stats AS (
        SELECT
            p.patient_id,
            p.first_name || ' ' || p.last_name         AS patient_name,
            p.insurance_type,
            COUNT(DISTINCT a.admission_id)              AS total_admissions,
            ROUND(AVG(
                JULIANDAY(a.discharge_date) - JULIANDAY(a.admission_date)
            ), 1)                                       AS avg_los,
            COUNT(DISTINCT d.icd_code)                  AS unique_diagnoses,
            MAX(CASE
                WHEN d.icd_code IN ('I50.9','J44.1','J18.9','I21.0','I21.9')
                THEN 1 ELSE 0
            END)                                        AS has_high_risk_dx
        FROM patients p
        JOIN admissions a ON p.patient_id = a.patient_id
        JOIN diagnoses d  ON a.patient_id = d.patient_id
        GROUP BY p.patient_id, p.first_name, p.last_name, p.insurance_type
    )
    SELECT *,
        CASE
            WHEN total_admissions > 1 AND has_high_risk_dx = 1 THEN 'High Risk'
            ELSE 'Standard'
        END AS risk_category
    FROM patient_stats
    ORDER BY total_admissions DESC
""")
 
fig, ax = plt.subplots(figsize=(11, 7))
 
colors = {'High Risk': '#D85A30', 'Standard': '#1D9E75'}
risk_df['color'] = risk_df['risk_category'].map(colors)
risk_df['bubble_size'] = risk_df['unique_diagnoses'] * 120
 
# Plot each patient
for _, row in risk_df.iterrows():
    ax.scatter(
        row['total_admissions'],
        row['avg_los'],
        s=row['bubble_size'],
        color=row['color'],
        alpha=0.75,
        edgecolors='white',
        linewidth=1.2,
        zorder=3
    )
 
# Label outliers — top right quadrant (high admissions AND high LOS)
avg_admits = risk_df['total_admissions'].median()
avg_los    = risk_df['avg_los'].median()
 
for _, row in risk_df.iterrows():
    if row['total_admissions'] >= avg_admits and row['avg_los'] >= avg_los:
        ax.annotate(
            row['patient_name'].split()[0],
            xy=(row['total_admissions'], row['avg_los']),
            xytext=(8, 4), textcoords='offset points',
            fontsize=8, color='#333333',
            fontweight='500'
        )
 
# Reference lines — median thresholds
ax.axvline(avg_admits, color='#AAAAAA', linestyle='--', linewidth=1, zorder=1)
ax.axhline(avg_los,    color='#AAAAAA', linestyle='--', linewidth=1, zorder=1)
 
# Quadrant labels
ax.text(avg_admits + 0.08, risk_df['avg_los'].max() * 0.97,
        'High admissions\nHigh LOS — Outliers',
        fontsize=8, color='#D85A30', fontstyle='italic', ha='left')
ax.text(0.72, risk_df['avg_los'].max() * 0.97,
        'Low admissions\nHigh LOS',
        fontsize=8, color='#888888', fontstyle='italic', ha='left')
 
# Bubble size legend
for n_diag, label in [(2, '2 diagnoses'), (4, '4 diagnoses'), (6, '6 diagnoses')]:
    ax.scatter([], [], s=n_diag * 120, color='#AAAAAA',
               alpha=0.6, label=label, edgecolors='white')
 
# Risk category legend
high_patch = mpatches.Patch(color='#D85A30', label='High Risk patient', alpha=0.75)
std_patch  = mpatches.Patch(color='#1D9E75', label='Standard patient',  alpha=0.75)
 
leg1 = ax.legend(handles=[high_patch, std_patch],
                 loc='lower right', fontsize=9,
                 title='Risk Category', title_fontsize=9,
                 framealpha=0.9)
ax.add_artist(leg1)
ax.legend(loc='lower left', fontsize=8,
          title='Bubble size = unique diagnoses',
          title_fontsize=8, framealpha=0.9)
 
ax.set_xlabel('Total Admissions', fontsize=11, labelpad=10)
ax.set_ylabel('Average Length of Stay (days)', fontsize=11, labelpad=10)
ax.set_title(
    'Patient Risk Profile — Admissions vs LOS vs Diagnosis Complexity',
    fontsize=13, fontweight='bold', pad=16
)
ax.spines[['top', 'right']].set_visible(False)
ax.set_xlim(0, risk_df['total_admissions'].max() + 1)
ax.set_ylim(0, risk_df['avg_los'].max() + 2)
ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))
 
plt.tight_layout()
plt.savefig('patient_risk_scatter.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved: patient_risk_scatter.png')
 
n_high = (risk_df['risk_category'] == 'High Risk').sum()
print(f'\nOutlier finding: {n_high} of {len(risk_df)} patients flagged as High Risk')
print('These patients have multiple admissions + a high-acuity primary diagnosis')
print('(Heart failure, COPD, Pneumonia, or Acute MI)')
 
 
```

```python
# ============================================================
# VISUALIZATION — LOS Box Plot by Department
# Shows spread and outliers within each department
# Proves you think beyond averages — a key analyst skill
# ============================================================
 
import matplotlib.pyplot as plt
import numpy as np
 
los_df = run("""
    SELECT
        department,
        ROUND(
            JULIANDAY(discharge_date) - JULIANDAY(admission_date)
        , 1) AS los_days,
        admission_type
    FROM admissions
    ORDER BY department
""")
 
# Prepare data per department
departments = los_df['department'].unique()
departments_sorted = (
    los_df.groupby('department')['los_days']
    .median()
    .sort_values(ascending=False)
    .index.tolist()
)
 
data_by_dept = [
    los_df[los_df['department'] == dept]['los_days'].tolist()
    for dept in departments_sorted
]
 
fig, ax = plt.subplots(figsize=(11, 6))
 
bp = ax.boxplot(
    data_by_dept,
    vert=True,
    patch_artist=True,
    widths=0.5,
    showfliers=True,
    medianprops=dict(color='white', linewidth=2),
    whiskerprops=dict(color='#666666', linewidth=1.2),
    capprops=dict(color='#666666', linewidth=1.2),
    flierprops=dict(
        marker='o', color='#D85A30',
        markerfacecolor='#D85A30',
        markersize=7, alpha=0.8,
        label='Outlier stay'
    )
)
 
# Color each box
box_colors = ['#1D9E75','#2AAFA0','#4DB6A8','#85CFC9','#B8E4E1']
for patch, color in zip(bp['boxes'], box_colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.8)
    patch.set_edgecolor('white')
 
# Add mean dots
means = [np.mean(d) for d in data_by_dept]
ax.scatter(range(1, len(departments_sorted) + 1), means,
           color='white', s=40, zorder=5,
           edgecolors='#1D4A40', linewidth=1.2,
           label='Mean LOS')
 
# Annotate outlier points
for i, data in enumerate(data_by_dept):
    arr = np.array(data)
    q1, q3 = np.percentile(arr, [25, 75])
    iqr = q3 - q1
    upper_fence = q3 + 1.5 * iqr
    outliers = arr[arr > upper_fence]
    for val in outliers:
        ax.annotate(
            f'{val:.0f} days',
            xy=(i + 1, val),
            xytext=(12, 0), textcoords='offset points',
            fontsize=8, color='#D85A30',
            fontweight='500',
            arrowprops=dict(arrowstyle='->', color='#D85A30', lw=0.8)
        )
 
# Formatting
ax.set_xticks(range(1, len(departments_sorted) + 1))
ax.set_xticklabels(departments_sorted, fontsize=10)
ax.set_ylabel('Length of Stay (days)', fontsize=11, labelpad=10)
ax.set_xlabel('Department', fontsize=11, labelpad=10)
ax.set_title(
    'Length of Stay Distribution by Department — Outlier Detection',
    fontsize=13, fontweight='bold', pad=16
)
ax.spines[['top', 'right']].set_visible(False)
ax.yaxis.grid(True, linestyle='--', alpha=0.5, color='#CCCCCC')
ax.set_axisbelow(True)
 
# Legend
from matplotlib.lines import Line2D
legend_elements = [
    mpatches.Patch(facecolor='#1D9E75', alpha=0.8, label='IQR (middle 50% of stays)'),
    Line2D([0], [0], marker='o', color='w', markerfacecolor='white',
           markeredgecolor='#1D4A40', markersize=7, label='Mean LOS'),
    Line2D([0], [0], marker='o', color='w', markerfacecolor='#D85A30',
           markersize=7, label='Outlier stay (>1.5× IQR)')
]
ax.legend(handles=legend_elements, fontsize=9, loc='upper right', framealpha=0.9)
 
# Insight annotation
ax.annotate(
    'Outliers represent unusually long stays\nthat may indicate complications or\ndischarge planning issues',
    xy=(0.01, 0.97), xycoords='axes fraction',
    fontsize=8, color='#555555', va='top',
    bbox=dict(boxstyle='round,pad=0.4', facecolor='#F5F5F5', edgecolor='#CCCCCC')
)
 
plt.tight_layout()
plt.savefig('los_outlier_boxplot.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved: los_outlier_boxplot.png')
 
# Print outlier findings
print('\n--- LOS OUTLIER FINDINGS ---')
for dept in departments_sorted:
    data = np.array(los_df[los_df['department'] == dept]['los_days'])
    q1, q3 = np.percentile(data, [25, 75])
    iqr = q3 - q1
    outliers = data[data > q3 + 1.5 * iqr]
    if len(outliers) > 0:
        print(f'  {dept}: {len(outliers)} outlier stay(s) — max {outliers.max():.0f} days '
              f'(avg for dept: {data.mean():.1f} days)')
 
```
